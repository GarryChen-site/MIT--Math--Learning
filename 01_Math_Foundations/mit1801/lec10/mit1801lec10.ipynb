```matlab
%% MIT 18.01 Lecture 10: Curve Sketching, Critical Points & Stability Curvature
% Mathematical Tool Application: Solves static equilibrium locations (f' = 0),
% classifies local stability via second-derivative curvature (f'' <=> Hessian),
% and identifies inflection bifurcation points in potential energy landscapes.


clear; clc; close all;


%% Part 1: Comprehensive Curve Analysis of Cubic System y = 3x - x^3
% Definition of system and derivatives
f_cubic = @(x) 3*x - x.^3;
df_cubic = @(x) 3 - 3*x.^2;
d2f_cubic = @(x) -6*x;


x_grid = linspace(-2.5, 2.5, 500);


figure('Color', 'w', 'Position', [100, 100, 1200, 520]);


% Subplot 1: Function f(x) with critical points, zeros, and inflection point
subplot(1, 2, 1);
plot(x_grid, f_cubic(x_grid), 'k-', 'LineWidth', 2.0, 'DisplayName', 'f(x) = 3x - x^3'); hold on;


% Critical Points: x = -1 (Min), x = 1 (Max)
plot(-1, f_cubic(-1), 'bo', 'MarkerSize', 8, 'MarkerFaceColor', 'b', ...
    'DisplayName', 'Local Min (-1, -2) [Stable Equilibrium]');
plot(1, f_cubic(1), 'ro', 'MarkerSize', 8, 'MarkerFaceColor', 'r', ...
    'DisplayName', 'Local Max (1, 2) [Unstable Equilibrium]');


% Inflection Point: x = 0
plot(0, f_cubic(0), 'ms', 'MarkerSize', 8, 'MarkerFaceColor', 'm', ...
    'DisplayName', 'Inflection Point (0, 0) [Zero Curvature]');


% Zeros: x = 0, +/- sqrt(3)
plot([-sqrt(3), sqrt(3)], [0, 0], 'k*', 'MarkerSize', 8, 'DisplayName', 'Zeros (x = \pm\surd 3)');


yline(0, 'k:', 'LineWidth', 1.0, 'HandleVisibility', 'off');
xline(0, 'k:', 'LineWidth', 1.0, 'HandleVisibility', 'off');
grid on;
xlim([-2.5, 2.5]); ylim([-4, 4]);
xlabel('State Variable x');
ylabel('Potential / Function Value y');
title('Cubic Landscape: Extrema & Concavity Inflection');
legend('Location', 'northwest');


%% Subplot 2: Derivative Slopes and Curvature Alignment
subplot(1, 2, 2);
plot(x_grid, df_cubic(x_grid), 'b--', 'LineWidth', 1.8, 'DisplayName', '1st Deriv f''(x) = 3(1 - x^2)'); hold on;
plot(x_grid, d2f_cubic(x_grid), 'm-.', 'LineWidth', 1.8, 'DisplayName', '2nd Deriv f''''(x) = -6x');
yline(0, 'k-', 'LineWidth', 1.2, 'HandleVisibility', 'off');


% Mark zero crossings of derivatives
plot([-1, 1], [0, 0], 'bo', 'MarkerFaceColor', 'b', 'DisplayName', 'Stationary Points (f'' = 0)');
plot(0, 0, 'ms', 'MarkerFaceColor', 'm', 'DisplayName', 'Inflection Root (f'''' = 0)');


grid on;
xlim([-2.5, 2.5]); ylim([-15, 15]);
xlabel('State Variable x');
ylabel('Derivative Magnitudes');
title('1st Deriv (Slope) and 2nd Deriv (Concavity Curvature)');
legend('Location', 'northeast');
```

```matlab
%% Part 2: Transcendental Function Optimization y = ln(x) / x
figure('Color', 'w', 'Position', [150, 150, 700, 450]);
x_trans = linspace(0.05, 10, 500);
y_trans = log(x_trans) ./ x_trans;


plot(x_trans, y_trans, 'b-', 'LineWidth', 2.0, 'DisplayName', 'f(x) = ln(x) / x'); hold on;
yline(0, 'k:', 'LineWidth', 1.2, 'DisplayName', 'Asymptote: y = 0 as x \rightarrow \infty');


% Critical Point at x = e, y = 1/e
x_crit_trans = exp(1);
y_crit_trans = 1 / exp(1);
plot(x_crit_trans, y_crit_trans, 'ro', 'MarkerSize', 8, 'MarkerFaceColor', 'r', ...
    'DisplayName', sprintf('Global Maximum: (e, 1/e) \\approx (%.3f, %.3f)', x_crit_trans, y_crit_trans));


% Zero crossing at x = 1
plot(1, 0, 'k*', 'MarkerSize', 8, 'DisplayName', 'Root: x = 1');


grid on;
xlim([0, 10]); ylim([-1.5, 0.6]);
xlabel('Independent Variable x (x > 0)');
ylabel('f(x)');
title('Transcendental Peak and Asymptotic Convergence');
legend('Location', 'southeast');
```
